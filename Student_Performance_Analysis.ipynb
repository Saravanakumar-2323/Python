**Project Title : Student Performance Analysis using Matplotlib**

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Student": ["Rahul", "Meera", "Vikram", "Ananya", "Dinesh", "Kavya", "Rohan", "Ishita"],
    "Math": [82, 71, 89, 76, 93, 68, 87, 95],
    "Science": [78, 75, 84, 72, 91, 80, 88, 97],
    "English": [85, 79, 81, 88, 86, 74, 90, 92],
    "Tamil": [73, 82, 77, 79, 89, 71, 85, 94],
    "Computer": [91, 88, 95, 83, 97, 86, 93, 99],
    "Study_Hours": [5, 4, 7, 6, 8, 3, 6, 9],
    "Attendance": [85, 78, 92, 80, 96, 72, 89, 98]
}

df = pd.DataFrame(data)
df["Total_Marks"] = df[["Math","Science","English","Tamil","Computer"]].sum(axis=1)

df
```


**Question 1**

How do marks change as study hours increase?


```python
plt.plot(df["Study_Hours"],df["Total_Marks"],marker="o")
plt.xlabel("Study Hours")
plt.ylabel("Total Marks")
plt.title("Marks vs Study Hours")
plt.show()
```

**Question 2**

Is Attendance related to marks scored?

```python
plt.scatter(df['Attendance'],df['Total_Marks'],marker="o")
plt.xlabel("Attendance")
plt.ylabel("Total Marks")
plt.title("Marks vs Attendance")
plt.show()
```

**Question 3**

Which Students Scored the highest and lowest marks?

```python
plt.bar(df['Student'],df['Total_Marks'])
plt.xlabel("Student")
plt.ylabel("Total Marks")
plt.title("Student Marks")
plt.show()
```

**Question 4**

How do marks vary across different subjects?

```python
subject = ["Math","Science","English","Tamil","Computer"]

plt.figure()
df[subject].mean().plot(kind = "bar")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.title("Average Marks per Subject")
plt.show()
```

**Question 5**

Compare Marks and Attendance trends together


```python
plt.plot(df["Student"],df["Total_Marks"],label = "Total Marks")
plt.plot(df["Student"],df["Attendance"],label = "Attendance")
plt.xlabel("Student")
plt.ylabel("Marks/Attendance")
plt.title("Student Marks and Attendance")
plt.legend()
plt.show()
```

**Question 6**

How does Performance improve step-by-step with study_hours?

```python
plt.step(df["Study_Hours"],df["Total_Marks"])
plt.xlabel("Study Hours")
plt.ylabel("Total Marks")
plt.title("Performance Improvement with Study Hours")
plt.show()
```
