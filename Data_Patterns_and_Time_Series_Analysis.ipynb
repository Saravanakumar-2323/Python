```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

**Question 1**

How does data behave whwn values grow exponentially?

```python
x = np.arange(1,11)
y = np.exp(x)
```

```python
plt.plot(x,y)
plt.yscale("log")
plt.xlabel("X Value")
plt.ylabel("Y Value")
plt.title("Exponential Growth")
plt.show()
```

**Question 2**

How do we visualize data containing both positive and negative values?

```python
values = [-1000, -200, -50, 0, 50, 200, 1000]
x = range(len(values))
```

```python
plt.plot(x, values)
plt.yscale("symlog")
plt.xlabel("X Value")
plt.ylabel("Y Value")
plt.title("Profit and Loss")
plt.show()
```

**Question 3**

How do we analyze probabilities bounded between 0 and 1?

```python
conversion_rate = [0.01, 0.05, 0.1, 0.3, 0.6, 0.9]
x = range(len(conversion_rate))
```

```python
plt.plot(x, conversion_rate, marker = 'o')
plt.yscale("logit")
plt.xlabel("X Value")
plt.ylabel("Y Value")
plt.title("Conversion Rate")
plt.grid(True)
plt.show()
```

**Question 4**

How can we visualize cyclical or circular data patterns?

```python
theta = np.linspace(0, 2*np.pi, 12)
demand = [20, 25, 30, 35, 40, 45, 50, 48, 42, 35, 28, 22]
```

```python
plt.figure(figsize = (6,6))
plt.polar(theta, demand, marker = 'o')
plt.title("Monthly Seasonal Demand")
plt.show()
```

**Question 5**

How do sales evolve over time?

```python
dates = pd.date_range(start = "2024-01-01", periods = 10, freq = "ME")
sales = [120, 135, 150, 160, 180, 200, 210, 230, 250, 270]

ts_df = pd.DataFrame({"Date": dates, "Sales": sales})
```

```python
plt.figure(figsize= (10,6))
plt.plot(ts_df["Date"], ts_df["Sales"], marker = 'o')
plt.xlabel("Date")
plt.ylabel("Sales")
plt.title("Time Series Sales Data")
plt.grid(True)
```

**Question 6**

How can we highlight trends in time-series data?

```python
plt.figure(figsize = (10,5))
plt.plot(ts_df["Date"], ts_df["Sales"], marker = 'o', label = "Sales", linestyle = "--")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.title("Time Series Sales Data")
plt.xticks(rotation = 45)
plt.show()
```
