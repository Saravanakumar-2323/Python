**Iterating Over array**

```python
a = np.arange(0,60,5).reshape(3,4)
print(a)
```

```python
for x in np.nditer(a):
    print(x)
```

**Iteration order(c-style and f-style)**

```python
print(a)
print()

for x in np.nditer(a,order='C'):
    print(x)
print()

for x in np.nditer(a,order='F'):
    print(x)
```

**Joining arrays**

```python
a = np.array([[1,2],[3,4]])
print("First Array: ")
print(a)

b = np.array([[5,6],[7,8]])
print("Second Array: ")
print(b)

print("Joining Array: ")
print(np.concatenate((a,b)))


print("Joining Array by axis 1: ")
print(np.concatenate((a,b),axis = 1))
```

**Resizing an array**

```python
a = np.array([[1,2,3],[4,5,6]])
print(a)
print(a.shape)
print()
b = np.resize(a,(3,2))
print(b)
print(b.shape)
print()
b = np.resize(a,(3,3))
print(b)
print(b.shape)
```

**Histogram**

```python
import matplotlib.pyplot as plt

a = np.array([20,87,4,40,53,74,56,51,11,20,40,15,79,25,27])

plt.hist(a, bins=[0,20,40,60,80,100])
plt.title("Histogram")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 3*np.pi, 0.1)
y = np.sin(x)

plt.plot(x, y)
plt.show()
```

```python
# Create a 6*6 two-dimensional array, and let 1 and 0 be placed alternatively across the diagonals

Z = np.zeros((6,6), dtype=int)

Z[1::2, ::2] = 1
Z[::2, 1::2] = 1

Z
```

```python
Z = np.random.rand(10,10)

Z[np.random.randint(10, size=5), np.random.randint(10, size=5)] = np.nan

Z
```

```python
print("Total number of missing values: \n", np.isnan(Z).sum())

print("Indexes of missing values: \n", np.argwhere(np.isnan(Z)))

inds = np.where(np.isnan(Z))
print(inds)

Z[inds] = 0

Z
```
