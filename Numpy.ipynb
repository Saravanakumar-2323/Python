**Array Manipulation - Changing Shape**

```python
import numpy as np

a = np.arange(9)
print("The Original Array: ")
print(a)
print()
b = a.reshape(3,3)
print("The modified Array: ")
print(b)
```

```python
print(b.flatten())
```

```python
b.flatten(order = "F")
```

```python
a = np.arange(12).reshape(3,4)
print(a)
```

```python
np.transpose(a)
```

```python
b = np.arange(8).reshape(2,4)
print(b)
```

```python
c = b.reshape(2,2,2)
print(c)
```

```python
np.rollaxis(c,2,1)
```

```python
np.swapaxes(c,2,1)
```

**Iterating Over array**

```python
a = np.arange(0,60,5).reshape(3,4)
print(a)
```

```python
for x in np.nditer(a):
    print(x)
```

**Iteration order(c-style and f-style)**

```python
print(a)
print()

for x in np.nditer(a,order='C'):
    print(x)
print()

for x in np.nditer(a,order='F'):
    print(x)
```

**Joining arrays**

```python
a = np.array([[1,2],[3,4]])
print("First Array: ")
print(a)

b = np.array([[5,6],[7,8]])
print("Second Array: ")
print(b)

print("Joining Array: ")
print(np.concatenate((a,b)))


print("Joining Array by axis 1: ")
print(np.concatenate((a,b),axis = 1))
```

**Splitting Array**

```python
np.split(a,[4,5])
```

**Resizing an array**

```python
a = np.array([[1,2,3],[4,5,6]])
print(a)
print(a.shape)
print()
b = np.resize(a,(3,2))
print(b)
print(b.shape)
print()
b = np.resize(a,(3,3))
print(b)
print(b.shape)
```

**Histogram**

```python
import matplotlib.pyplot as plt

a = np.array([20,87,4,40,53,74,56,51,11,20,40,15,79,25,27])

plt.hist(a, bins=[0,20,40,60,80,100])
plt.title("Histogram")
plt.show()
```
