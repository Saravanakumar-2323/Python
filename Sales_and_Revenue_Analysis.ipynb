```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Electronics": [120, 150, 180, 200, 220, 250],
    "Clothing": [100, 110, 130, 160, 180, 210],
    "Groceries": [90, 95, 100, 110, 120, 130],
    "Furniture": [60, 70, 80, 90, 100, 120]
}

df = pd.DataFrame(data)

df
```

**Question 1**

How do monthly sales trends differ across product categories?

```python
plt.figure(figsize = (12,8))

plt.subplot(2,2,1)
plt.plot(df["Month"], df["Electronics"])
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Electronic Sales")

plt.subplot(2,2,2)
plt.plot(df["Month"], df["Clothing"])
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Clothing Sales")

plt.subplot(2,2,3)
plt.plot(df["Month"], df["Groceries"])
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Groceries Sales")

plt.subplot(2,2,4)
plt.plot(df["Month"], df["Furniture"])
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Furniture Sales")

plt.tight_layout()
plt.show()
```

**Question 2**

Compare all products sales in single view visualization

```python
plt.figure(figsize = (8,6))

plt.plot(df["Month"], df["Electronics"], label = "Electronics")
plt.plot(df["Month"], df["Clothing"], label = "Clothing")
plt.plot(df["Month"], df["Groceries"], label = "Groceries")
plt.plot(df["Month"], df["Furniture"], label = "Furniture")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Product Sales Comparison")
plt.legend()
plt.grid(True)
plt.show()
```

**Question 3**

What is the sales distribution for Electronics?

```python
plt.figure(figsize = (6,4))
plt.hist(df["Electronics"], bins = 5, color = "green", edgecolor = "black")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.title("Electronics Sales Distribution")
plt.show()
```

**Question 4**

Which product category has the highest total sales?

```python
total_sales = df.iloc[:,1:].sum()

plt.figure(figsize = (6,4))
plt.bar(total_sales.index, total_sales.values, color = "magenta")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.title("Total Sales by Product Category")
plt.show()
```

**Question 5**

What is the percentage contribution of each product to total sales?

```python
plt.figure(figsize = (6,4))
plt.pie(total_sales, labels = total_sales.index, autopct = "%1.2f%%", explode = (0.1,0,0,0))
plt.title("Percentage Contribution  of Products to Total Sales")
plt.show()
```

**Question 6**

How does cumulative sales grow over time?

```python
plt.figure(figsize = (6,4))

plt.stackplot(df["Month"], df["Electronics"], df["Clothing"], df["Groceries"], df["Furniture"], labels = ["Electronics", "Clothing", "Groceries", "Furniture"])
plt.xlabel("Month")
plt.ylabel("Cumulative Sales")
plt.title("Cumulative Sales Growth Over Time")
plt.legend()
plt.show()
```
