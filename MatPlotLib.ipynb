## Line and SETP Methods

#### Recap of the line plot

```python
x = np.arange(10)
y1 = [1,9,7,10,3,16,2,20,5,22]
y2 = [11,22,7,1,14,6,8,2,15,3]
plt.plot(x,y1,'--r',linewidth = 4)
plt.plot(x,y2,'--y',linewidth = 3)
plt.show()
```

#### Assign plots to lines variable and apply individual settings

```python
x = np.arange(10)
y1 = [1,9,7,10,3,16,2,20,5,22]
y2 = [11,22,7,1,14,6,8,2,15,3]
lines = plt.plot(x,y1,'--y',x,y2,'-r')
plt.show()
```

#### Use SETP method to apply bulk settings using single statement

```python
x = np.arange(10)
y1 = [1,9,7,10,3,16,2,20,5,22]
y2 = [11,22,7,1,14,6,8,2,15,3]
lines = plt.plot(x,y1,x,y2)
plt.setp(lines, color='r',linewidth=3)
plt.show()
```

## Figure and Subplots

#### Creating two figures without any data

```python
t1 = np.arange(0.0,5.0,0.1)
y1 = np.sin(2*np.pi*t1)
y2 = np.cos(2*np.pi*t1)

plt.figure(1)
plt.figure(2)

plt.show()
```

#### Filling data in these subplots

```python
t1 = np.arange(0.0,5.0,0.1)
y1 = np.sin(2*np.pi*t1)
y2 = np.cos(2*np.pi*t1)

plt.figure(1)
plt.subplot(121)
plt.title("Plot 1")
plt.plot(t1,y1,'r')
plt.subplot(122)
plt.title("Plot 2")
plt.plot(t1,y2,'g')

plt.figure(2)
plt.subplot(121)
plt.title("Plot 3")
plt.plot(t1,y2,'k')
plt.subplot(122)
plt.title("Plot 4")
plt.plot(t1,y1,'m')

plt.show()


```

## Display Grid in Plot


```python
t1 = np.arange(0.0,5.0,0.1)
y1 = np.sin(2*np.pi*t1)
y2 = np.cos(2*np.pi*t1)

plt.figure(1)
plt.subplot(121)
plt.title("Plot 1")
plt.plot(t1,y1,'r')
plt.subplot(122)
plt.title("Plot 2")
plt.plot(t1,y2,'g')
plt.grid(True)

plt.figure(2)
plt.subplot(121)
plt.title("Plot 3")
plt.plot(t1,y2,'k')
plt.grid(True)
plt.subplot(122)
plt.title("Plot 4")
plt.plot(t1,y1,'m')

plt.show()


```

## Histogram

#### Simplest way to create a histogram

```python
data = np.random.randn(1000)
h1 = 0.50*data+4
plt.hist(h1)
plt.title("Histogram Plot for Random Data",fontsize = 14)
plt.show()
```

#### Apply bins, alignment and color to histogram

```python
data = np.random.randn(1000)
h1 = 0.50*data+4
plt.hist(h1, 5, align = 'mid', color = 'g')
plt.title("Histogram plot for Random Data", fontsize = 14)
plt.show()
```

#### Change Histogram Orientation

```python
data = np.random.randn(1000)
h1 = 0.50*data+4
plt.hist(h1, 5, align = 'mid', color = 'm',orientation = 'horizontal')
plt.title("Histogram plot for Random Data", fontsize = 14)
plt.show()
```

#### Apply Histogram Display Type and Bottom Threshold


```python
data = np.random.randn(1000)
h1 = 0.50*data+4
plt.hist(h1, 5, align = 'mid', color = 'r', histtype = 'step',bottom = 100)
plt.title("Histogram plot for Random Data", fontsize = 14)
plt.show()
```

#### Displaying Values over Log Scale


```python
data = np.random.randn(1000)
h1 = 0.50*data+4
plt.hist(h1, 5, align = 'mid', color = 'c', histtype = 'bar',bottom = 100, log=True)
plt.title("Histogram plot for Random Data", fontsize = 14)
plt.show()
```

## Stack Plot

#### Plotting basic Stack Plot

```python
avg_score = [60,70,75,25,64,45,55,46,36,52]
sub = ['Sci','Phy','Che','Math','Bio','IT','Tech','Py','ECE','CS']
grace_mark = [12,24,6,5,7,10,7,20,4,9]

plt.title("Average Score for Subjects",fontsize = 16)
plt.xlabel("Subjects", fontsize = 14, color = 'r')
plt.ylabel("Scores", fontsize = 14, color = 'm')

plt.bar(sub, avg_score, label = 'Average Score')
plt.bar(sub, grace_mark, bottom = avg_score)

plt.show()

```

## Non - Linear Plotting


#### Generating sample data for Plotting


```python
t = np.arange(0.01, 20.0, 0.01)
y1 = np.exp(-t / 5.0)
y2 = np.sin(2 * np.pi * t)
plt.plot(t, y1, 'v-r', t, y2, '-g')
plt.show()
```

#### Normal Plot y1 over t

```python
plt.plot(y1,t)
plt.title("Normal plot of y1 over t", fontsize = 14)
plt.xlabel("Time")
plt.ylabel("Data is y1")
plt.show()
```

#### Plotting y1 over t | But making Y axis as logarthmic(base 10)

```python
plt.semilogy(t, y1)
plt.title("Y Axis as Logarthmic axis", fontsize = 14)
plt.xlabel("Time")
plt.ylabel("Data is y2")
plt.show()
```

#### Plotting y1 over t | But making X axis as logarthmic(base 10)

```python
plt.semilogx(t, np.sin(2 * np.pi * t))
plt.title("X Axis as Logarthmic axis", fontsize = 14)
plt.xlabel("Time")
plt.ylabel("Data is y2")
plt.show()
```

#### Plotting y1 over t | But making X axis as logarthmic(base 2)

```python
plt.semilogx(t, 20 * y1, base= 2)
plt.title("Y Axis as Logarthmic axis with base 2", fontsize = 14)
plt.xlabel("Time")
plt.ylabel("Data is y1")
plt.show()
```

## Symlog and logit plot 


#### Generate data and customize it

```python
y = np.random.normal(loc = 0.5, scale = 0.4, size = 1000)
y[(y > 0) & (y < 1)]
y.sort()

x = np.arange(len(y))
plt.plot(x,y)
```

#### Plotting symmetric log | Y Scale 

```python
plt.plot(x, (y - y.mean()))
plt.yscale('symlog', linthresh = 0.01)
plt.title("This is the Symmetric Log", fontsize = 14)
plt.show()
```

#### Plotting on logit scale

```python
plt.plot(x,y)
plt.yscale('logit')
plt.title("A logit plot", fontsize = 14)
plt.show()
```
