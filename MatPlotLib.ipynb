## Plotting Categorical Data

#### Plot city population vs age group as a category

```python
age_group = ['<18','18-30','31-50','51-70','>70']
num = [12,18,17,8,6]
plt.bar(age_group, num)
plt.xlabel("Age Group")
plt.ylabel("Population x100000")
plt.title("Population Stats of India")
plt.show()
```

```python
age_group = ['<18','18-30','31-50','51-70','>70']
num = [12,18,17,8,6]
colors = ['r','b','g','y','c']
plt.bar(age_group, num, color = colors, width = 0.5)
plt.xlabel("Age Group")
plt.ylabel("Population x100000")
plt.title("Population Stats of India")
plt.show()
```

## Line and SETP Methods

#### Recap of the line plot

```python
x = np.arange(10)
y1 = [1,9,7,10,3,16,2,20,5,22]
y2 = [11,22,7,1,14,6,8,2,15,3]
plt.plot(x,y1,'--r',linewidth = 4)
plt.plot(x,y2,'--y',linewidth = 3)
plt.show()
```

#### Assign plots to lines variable and apply individual settings

```python
x = np.arange(10)
y1 = [1,9,7,10,3,16,2,20,5,22]
y2 = [11,22,7,1,14,6,8,2,15,3]
lines = plt.plot(x,y1,'--y',x,y2,'-r')
plt.show()
```

#### Use SETP method to apply bulk settings using single statement

```python
x = np.arange(10)
y1 = [1,9,7,10,3,16,2,20,5,22]
y2 = [11,22,7,1,14,6,8,2,15,3]
lines = plt.plot(x,y1,x,y2)
plt.setp(lines, color='r',linewidth=3)
plt.show()
```

## Figure and Subplots

#### Creating two figures without any data

```python
t1 = np.arange(0.0,5.0,0.1)
y1 = np.sin(2*np.pi*t1)
y2 = np.cos(2*np.pi*t1)

plt.figure(1)
plt.figure(2)

plt.show()
```

#### Filling data in these subplots

```python
t1 = np.arange(0.0,5.0,0.1)
y1 = np.sin(2*np.pi*t1)
y2 = np.cos(2*np.pi*t1)

plt.figure(1)
plt.subplot(121)
plt.title("Plot 1")
plt.plot(t1,y1,'r')
plt.subplot(122)
plt.title("Plot 2")
plt.plot(t1,y2,'g')

plt.figure(2)
plt.subplot(121)
plt.title("Plot 3")
plt.plot(t1,y2,'k')
plt.subplot(122)
plt.title("Plot 4")
plt.plot(t1,y1,'m')

plt.show()


```

## Display Grid in Plot


```python
t1 = np.arange(0.0,5.0,0.1)
y1 = np.sin(2*np.pi*t1)
y2 = np.cos(2*np.pi*t1)

plt.figure(1)
plt.subplot(121)
plt.title("Plot 1")
plt.plot(t1,y1,'r')
plt.subplot(122)
plt.title("Plot 2")
plt.plot(t1,y2,'g')
plt.grid(True)

plt.figure(2)
plt.subplot(121)
plt.title("Plot 3")
plt.plot(t1,y2,'k')
plt.grid(True)
plt.subplot(122)
plt.title("Plot 4")
plt.plot(t1,y1,'m')

plt.show()


```

## Histogram

#### Simplest way to create a histogram

```python
data = np.random.randn(1000)
h1 = 0.50*data+4
plt.hist(h1)
plt.title("Histogram Plot for Random Data",fontsize = 14)
plt.show()
```

#### Apply bins, alignment and color to histogram

```python
data = np.random.randn(1000)
h1 = 0.50*data+4
plt.hist(h1, 5, align = 'mid', color = 'g')
plt.title("Histogram plot for Random Data", fontsize = 14)
plt.show()
```

#### Change Histogram Orientation

```python
data = np.random.randn(1000)
h1 = 0.50*data+4
plt.hist(h1, 5, align = 'mid', color = 'm',orientation = 'horizontal')
plt.title("Histogram plot for Random Data", fontsize = 14)
plt.show()
```

#### Apply Histogram Display Type and Bottom Threshold


```python
data = np.random.randn(1000)
h1 = 0.50*data+4
plt.hist(h1, 5, align = 'mid', color = 'r', histtype = 'step',bottom = 100)
plt.title("Histogram plot for Random Data", fontsize = 14)
plt.show()
```

#### Displaying Values over Log Scale


```python
data = np.random.randn(1000)
h1 = 0.50*data+4
plt.hist(h1, 5, align = 'mid', color = 'c', histtype = 'bar',bottom = 100, log=True)
plt.title("Histogram plot for Random Data", fontsize = 14)
plt.show()
```

## Pie Chart

#### Simplest way to create a Pie chart

```python
percent_result = [17,30,27,18,8]
sub = ['Math','Che','Phy','Bio','CS']

plt.pie(percent_result, labels = sub)
plt.show()
```

#### Poping out of Focusing Regions of Pie

```python
percent_result = [17,30,27,18,8]
sub = ['Math','Che','Phy','Bio','CS']

plt.pie(percent_result, explode = (0,0,0.1,0,0),labels = sub)
plt.show()
```

#### Displaying Shadow in pie chart

```python
percent_result = [17,30,27,18,8]
sub = ['Math','Che','Phy','Bio','CS']

plt.pie(percent_result, explode = (0,0,0.1,0,0),labels = sub, shadow = True)
plt.show()
```

#### Displaying Portion Percentage on each part

```python
percent_result = [17,30,27,18,40]
sub = ['Math','Che','Phy','Bio','CS']

plt.pie(percent_result, explode = (0,0,0.1,0,0),labels = sub, autopct = '%1.2f%%')
plt.show()
```

## Stack Plot

#### Plotting basic Stack Plot

```python
avg_score = [60,70,75,25,64,45,55,46,36,52]
sub = ['Sci','Phy','Che','Math','Bio','IT','Tech','Py','ECE','CS']
grace_mark = [12,24,6,5,7,10,7,20,4,9]

plt.title("Average Score for Subjects",fontsize = 16)
plt.xlabel("Subjects", fontsize = 14, color = 'r')
plt.ylabel("Scores", fontsize = 14, color = 'm')

plt.bar(sub, avg_score, label = 'Average Score')
plt.bar(sub, grace_mark, bottom = avg_score)

plt.show()

```

#### Changing colors and width for each bar

```python
avg_score = [60,70,75,25,64,45,55,46,36,52]
sub = ['Sci','Phy','Che','Math','Bio','IT','Tech','Py','ECE','CS']
grace_mark = [12,24,6,5,7,10,7,20,4,9]

plt.title("Average Score for Subjects",fontsize = 16)
plt.xlabel("Subjects", fontsize = 14, color = 'r')
plt.ylabel("Scores", fontsize = 14, color = 'm')
colors = ['blue','orange','magenta','red','yellow','red','purple','black','cyan','orange']

plt.bar(sub, avg_score, label = 'Average Score',width = 0.7, color = colors)
plt.bar(sub, grace_mark, bottom = avg_score, width = 0.7, color = 'green')

plt.show()

```

#### Adding Legends to the Stack Plot


```python
avg_score = [60,70,75,25,64,45,55,46,36,52]
sub = ['Sci','Phy','Che','Math','Bio','IT','Tech','Py','ECE','CS']
grace_mark = [12,24,6,5,7,10,7,20,4,9]

plt.title("Average Score for Subjects",fontsize = 16)
plt.xlabel("Subjects", fontsize = 14, color = 'r')
plt.ylabel("Scores", fontsize = 14, color = 'm')
colors = ['blue','orange','magenta','red','yellow','red','purple','black','cyan','orange']

plt.bar(sub, avg_score, label = 'Average Score',width = 0.7, color = colors)
plt.bar(sub, grace_mark, bottom = avg_score, width = 0.7, color = 'green', label = 'Grace Mark')

plt.legend()
plt.show()

```

#### Adding error or Deviations in the stack

```python
avg_score = [60,70,75,25,64,45,55,46,36,52]
sub = ['Sci','Phy','Che','Math','Bio','IT','Tech','Py','ECE','CS']
grace_mark = [12,24,6,5,7,10,7,20,4,9]

plt.title("Average Score for Subjects",fontsize = 16)
plt.xlabel("Subjects", fontsize = 14, color = 'r')
plt.ylabel("Scores", fontsize = 14, color = 'm')
colors = ['blue','orange','magenta','red','yellow','red','purple','black','cyan','orange']

plt.bar(sub, avg_score, label = 'Average Score',width = 0.7, color = colors,  yerr = grace_mark)
plt.bar(sub, grace_mark, bottom = avg_score, width = 0.7, color = 'green', label = 'Grace Mark')

plt.legend()
plt.show()

```
