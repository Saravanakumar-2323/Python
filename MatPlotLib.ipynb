## Specify a Common X-Axis and assign separate markers for both

```python
x = np.arange(100)
y1 = np.random.randn(100)
y2 = np.random.randn(100)
plt.scatter(x, y1, marker = 'o', label = "y1 with x")
plt.scatter(x, y1, marker = 'v', label = "y2 with x")
plt.legend(loc = 'upper left')
plt.show()
```

```python
x = np.arange(100)
y1 = np.random.randn(100)
y2 = np.random.randn(100)
plt.scatter(x, y1, marker = 'o', c = 'r')
plt.scatter(x, y2, marker = 'v', c = 'y')
plt.show()
```

## Plotting Categorical Data

#### Plot city population vs age group as a category

```python
age_group = ['<18','18-30','31-50','51-70','>70']
num = [12,18,17,8,6]
plt.bar(age_group, num)
plt.xlabel("Age Group")
plt.ylabel("Population x100000")
plt.title("Population Stats of India")
plt.show()
```

#### Applying various bar properties like color , width , edgecolor

```python
age_group = ['<18','18-30','31-50','51-70','>70']
num = [12,18,17,8,6]
plt.bar(age_group, num, color = 'b', edgecolor = 'm', width = 0.5)
plt.xlabel("Age Group")
plt.ylabel("Population x100000")
plt.title("Population Stats of India")
plt.show()
```

```python
age_group = ['<18','18-30','31-50','51-70','>70']
num = [12,18,17,8,6]
colors = ['r','b','g','y','c']
plt.bar(age_group, num, color = colors, width = 0.5)
plt.xlabel("Age Group")
plt.ylabel("Population x100000")
plt.title("Population Stats of India")
plt.show()
```

## Line and SETP Methods

#### Recap of the line plot

```python
x = np.arange(10)
y1 = [1,9,7,10,3,16,2,20,5,22]
y2 = [11,22,7,1,14,6,8,2,15,3]
plt.plot(x,y1,'--r',linewidth = 4)
plt.plot(x,y2,'--y',linewidth = 3)
plt.show()
```

#### Assign plots to lines variable and apply individual settings

```python
x = np.arange(10)
y1 = [1,9,7,10,3,16,2,20,5,22]
y2 = [11,22,7,1,14,6,8,2,15,3]
lines = plt.plot(x,y1,'--y',x,y2,'-r')
plt.show()
```

#### Use SETP method to apply bulk settings using single statement

```python
x = np.arange(10)
y1 = [1,9,7,10,3,16,2,20,5,22]
y2 = [11,22,7,1,14,6,8,2,15,3]
lines = plt.plot(x,y1,x,y2)
plt.setp(lines, color='r',linewidth=3)
plt.show()
```
