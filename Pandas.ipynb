## Visualization

### Create Basic Plots in Pandas

```python
import matplotlib.pyplot as plt
```

```python
df_day.plot(x = 'dteday',y = 'cnt', kind = 'line', figsize = (10,5), title = "Bike Rentals Over Time")
plt.xlabel("Date")
plt.ylabel("Total Rentals")
plt.show
```

```python
df_day['cnt'].hist(bins = 20, color = 'lightgreen', edgecolor = 'black',figsize = (10,5))
plt.title("Distribution of Rentals")
plt.xlabel("Rental Count")
plt.ylabel("Frequency")
plt.grid(False)
plt.show()
```

```python
df_day.boxplot(column = ['cnt'], grid = False, figsize = (5,5))
plt.title("Box Plot of Total Bike Remtals")
plt.show()
```

```python
df_season = df_day.groupby('season')['cnt'].mean()
df_season.plot(kind = 'line', marker = 'o', color = 'red', figsize = (8, 5), title = "Average Rentals by Season")
plt.xlabel("Season (1: Spring, 2: Summer, 3: Fall, 4: Winter)")
plt.ylabel("Average Rentals")
plt.show()
```

```python
df_season.plot(kind = 'bar',color = ['blue', 'orange', 'green', 'red'], figsize = (8,5), title = "Total Rentals by Season")
plt.xlabel("Season")
plt.ylabel("Total Rentals")
plt.xticks(rotation = 0)
plt.show()
```

```python
plt.figure(figsize = (10,5))

plt.plot(df_day['dteday'], df_day['cnt'],color = 'orange', linestyle = '--', marker = 'o', markersize = 4)
plt.title("Customized Line PLot: Bike Rentals Over Time", fontsize = 14)
plt.xlabel("Date", fontsize = 12)
plt.ylabel("Total Rentals", fontsize = 12)
plt.show()
```

```python
df_day.plot(x = 'temp', y = 'cnt', kind = 'scatter', color = 'purple', alpha = 0.5, title = "Rentals vs Temperature")
plt.xlabel("Temperature (Normalized)")
plt.ylabel("Total Rentals")
plt.show()
```
